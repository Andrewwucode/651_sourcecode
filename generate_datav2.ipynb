```python
import numpy as np
import pandas as pd

# Random number generator
rng = np.random.default_rng(42)

# Number of simulated customers
n = 100

# Possible customer comments based on different categories of experiences
comments = {
    "Long Wait": [
        "We waited almost an hour for a table.",
        "The wait was too long.",
        "The restaurant was very busy and seating took forever.",
        "Great food, but the wait time was frustrating.",
    ],
    "Positive": [
        "The seafood was delicious and the staff was friendly.",
        "We were seated quickly and had a great experience.",
        "Excellent food and great service.",
        "Everything was amazing. I would come back.",
    ],
    "Slow Service": [
        "Our server took too long to take our order.",
        "The food took a long time to arrive.",
        "We waited too long for our drinks.",
    ],
    "Neutral": [
        "The food was okay and the service was average.",
        "The experience was fine, nothing special.",
        "The restaurant was crowded but the food was decent.",
    ],
}

records = []

for i in range(n):

    money_spent = np.clip(rng.normal(loc=80, scale=60), 4.50, None).round(2)
    # Simulate meal type, p is the probability distribution for each meal type
    meal = rng.choice(["Lunch", "Dinner"], p=[0.35, 0.65])

    # Simulate wait time in minutes, p is the probability distribution for each wait time option
    wait_time = int(
        rng.choice(
            [5, 10, 15, 20, 30, 45, 60, 75],
            p=[0.05, 0.10, 0.15, 0.20, 0.20, 0.15, 0.10, 0.05],
        )
    )

    # Simulate food quality rating (1-5)
    food_rating = int(rng.choice([1, 2, 3, 4, 5], p=[0.05, 0.10, 0.20, 0.35, 0.30]))

    # Simulate service rating (1-5)
    service_rating = int(rng.choice([1, 2, 3, 4, 5], p=[0.05, 0.15, 0.25, 0.35, 0.20]))

    # Wait satisfaction decreases as wait time increases
    wait_satisfaction = int(
        np.clip(5 - wait_time / 15 + rng.normal(0, 0.7), 1, 5).round()
    )

    # Overall satisfaction depends on multiple factors
    overall_satisfaction = int(np.clip(0.4 * food_rating + 0.3 * service_rating + 0.3 * wait_satisfaction, 1, 5).round())

    # Select a comment based on the experience
    if wait_time >= 45:
        category = "Long Wait"
    elif service_rating <= 2:
        category = "Slow Service"
    elif overall_satisfaction >= 4:
        category = "Positive"
    else:
        category = "Neutral"

    comment = rng.choice(comments[category])

    records.append(
        {
            "customer_id": i + 1,
            "meal": meal,
            "wait_time_minutes": wait_time,
            "food_rating": food_rating,
            "service_rating": service_rating,
            "wait_satisfaction": wait_satisfaction,
            "overall_satisfaction": overall_satisfaction,
            "money_spent": money_spent,
            "comment": comment,
            "VOC_Theme": category,
        }
    )

# Convert records to a pandas DataFrame
df = pd.DataFrame(records)

# Export to Excel
# df.to_excel("boiling_crab_survey.xlsx", index=False)

# Preview the first 10 rows
df.head(50).sort_values(by="money_spent", ascending=False)

# Display summary statistics for the DataFrame
df.describe()
df["comment"].value_counts()
df["VOC_Theme"].value_counts()

```
